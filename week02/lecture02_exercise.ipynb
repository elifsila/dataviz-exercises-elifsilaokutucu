## Setup — Run this cell first


```python
import os
import pandas as pd
import numpy as np

# World Happiness Report 2023 — representative data
# Source: https://www.kaggle.com/datasets/ajaypalsinghlo/world-happiness-report-2023

# Works both in the repo (../data/) and in Google Colab (upload prompt)
DATA_PATH = '../data/world_happiness_2023.csv'
if not os.path.exists(DATA_PATH):
    DATA_PATH = 'world_happiness_2023.csv'
    if not os.path.exists(DATA_PATH):
        from google.colab import files
        uploaded = files.upload()          # choose world_happiness_2023.csv
        DATA_PATH = list(uploaded.keys())[0]

df = pd.read_csv(DATA_PATH)
df.columns = ['Country','Region','Happiness_Score','GDP','Social_Support',
              'Life_Expectancy','Freedom','Generosity','Corruption']


print(f"Dataset: {len(df)} countries, {len(df.columns)} columns")
print(df.head())
```

```python
import plotly.express as px
import plotly.graph_objects as go

# Explore the dataset before you start
print("Regions in dataset:")
print(df['Region'].value_counts())
print("\nScore range:", df['Happiness_Score'].min(), "–", df['Happiness_Score'].max())
print("\nBottom 10 countries:")
print(df.nsmallest(10, 'Happiness_Score')[['Country','Region','Happiness_Score']])
```

## Task 1 — Regional Comparison Bar Chart

**What to build:** A horizontal bar chart showing the **average happiness score by region**, sorted from highest to lowest.

**Requirements:**
- Horizontal orientation (category names are long)
- Sorted by score, descending (so the happiest region is at the top)
- Zero baseline on x-axis
- At least one design choice that goes beyond the Plotly default (colour, annotation, labels, etc.)
- An insight title that answers: *which region stands out and why does it matter?*

**Hint:** Use `df.groupby('Region')['Happiness_Score'].mean()` to compute the averages.


```python
# Task 1: Regional comparison bar chart
# -------------------------------------

# Step 1: Compute average happiness score by region
region_avg = (df.groupby('Region')['Happiness_Score']
              .mean()
              .reset_index()
              .sort_values('Happiness_Score'))  # sort for horizontal bar

print(region_avg)

# Step 2: Build the chart
top_region = region_avg.iloc[-1]      # sorted ascending -> last row = happiest
bottom_region = region_avg.iloc[0]
gap = top_region['Happiness_Score'] - bottom_region['Happiness_Score']

# Grey everything, highlight only the leading region
region_avg['highlight'] = np.where(region_avg['Region'] == top_region['Region'], 'Top', 'Other')

fig = px.bar(
    region_avg,
    x='Happiness_Score', y='Region',
    orientation='h',                                  # Rule 3: horizontal for long labels
    color='highlight',
    color_discrete_map={'Top': '#2E75B6', 'Other': '#D0D0D0'},
    text='Happiness_Score',
    title=(f"{top_region['Region']} is the happiest region — "
           f"{gap:.1f} points ahead of {bottom_region['Region']}"),
    labels={'Happiness_Score': 'Average Happiness Score (0–10)', 'Region': ''},
    height=500
)

fig.update_traces(marker_line_width=0, texttemplate='%{text:.2f}', textposition='outside')

fig.update_layout(
    xaxis=dict(range=[0, 8.5], gridcolor='#EEEEEE'),  # Rule 1: zero baseline
    yaxis=dict(gridcolor='white',
               categoryorder='array',
               categoryarray=region_avg['Region'].tolist()),  # Rule 2: keep sorted order
    plot_bgcolor='white',
    paper_bgcolor='white',
    font=dict(family='Arial', size=13),
    margin=dict(l=10, r=30, t=60, b=40),
    showlegend=False
)
fig.show()
```

## Task 2 — Bottom vs. Top: A Contrast Story

**What to build:** A bar chart that highlights the **gap between the happiest and least happy countries**, focusing on a specific insight.

**Requirements:**
- Show the **top 8 AND bottom 8 countries** together (16 bars total)
- Use **colour** to distinguish the two groups (not Plotly's default rainbow)
- Add a **visual separator or annotation** that emphasises the gap
- Insight title that tells the story of the gap

**Hint:** Use `pd.concat([df.nlargest(8,'Happiness_Score'), df.nsmallest(8,'Happiness_Score')])` to get both groups.

**Stretch goal:** Add a vertical reference line showing the global average.


```python
# Task 2: Top 8 vs. Bottom 8 contrast
# ------------------------------------

# Step 1: Get top and bottom countries
top8 = df.nlargest(8, 'Happiness_Score').copy()
top8['Group'] = 'Top 8'
bottom8 = df.nsmallest(8, 'Happiness_Score').copy()
bottom8['Group'] = 'Bottom 8'

combined = pd.concat([bottom8, top8]).sort_values('Happiness_Score')
global_avg = df['Happiness_Score'].mean()
print(f"Global average: {global_avg:.2f}")

# Step 2: Build the chart
top_mean = top8['Happiness_Score'].mean()
bottom_mean = bottom8['Happiness_Score'].mean()
gap = top_mean - bottom_mean

fig = px.bar(
    combined,
    x='Happiness_Score', y='Country',
    orientation='h',
    color='Group',
    color_discrete_map={'Top 8': '#2E75B6', 'Bottom 8': '#E63946'},
    text='Happiness_Score',
    title=(f"The 8 happiest countries score {top_mean / bottom_mean:.1f}x higher than the 8 least happy — "
           f"a {gap:.1f}-point gap"),
    labels={'Happiness_Score': 'Happiness Score (0–10)', 'Country': '', 'Group': ''},
    height=650)

fig.update_traces(marker_line_width=0, texttemplate='%{text:.2f}', textposition='outside')

# Visual separator between the two groups (bars 0-7 = bottom, 8-15 = top)
fig.add_hline(y=7.5, line_dash='dot', line_color='#999999', line_width=1)

# Stretch: global average reference line
fig.add_vline(
    x=global_avg, line_dash='dash', line_color='#333333', line_width=1.5,
    annotation_text=f'Global average: {global_avg:.2f}',
    annotation_position='top',
    annotation_font=dict(size=11, color='#333333'))

# Annotation emphasising the gap
fig.add_annotation(
    x=global_avg, y=7.5, xanchor='left', xshift=12,
    text=f'<b>{gap:.1f}-point gap</b><br>between group averages',
    showarrow=False, align='left',
    font=dict(size=12, color='#333333'),
    bgcolor='white', bordercolor='#999999', borderwidth=1, borderpad=4)

fig.update_layout(
    xaxis=dict(range=[0, 8.5], gridcolor='#EEEEEE'),  # zero baseline
    yaxis=dict(gridcolor='white',
               categoryorder='array',
               categoryarray=combined['Country'].tolist()),
    plot_bgcolor='white',
    paper_bgcolor='white',
    font=dict(family='Arial', size=13),
    legend=dict(orientation='h', y=1.06, x=0, traceorder='reversed'),
    margin=dict(l=10, r=30, t=90, b=40))
fig.show()
```

## Done? Stretch Goal

If you finish both tasks with time to spare, try this:

**Task 3 (stretch):** Build a **grouped bar chart** comparing 2 sub-factors (e.g. `GDP_per_capita` and `Freedom`) across the 5 most populated regions. Use colour meaningfully and write an insight title.

Regions to include: `'Western Europe'`, `'Latin America'`, `'East Asia'`, `'Sub-Saharan Africa'`, `'South Asia'`


```python
# Stretch goal — grouped bar chart
# Region names can differ slightly in the CSV, so match by keyword
keywords = ['Western Europe', 'Latin America', 'East Asia', 'Sub-Saharan Africa', 'South Asia']
mask = df['Region'].apply(lambda r: any(r.startswith(k) for k in keywords))

factors = (df[mask].groupby('Region')[['GDP', 'Freedom']]
           .mean()
           .reset_index()
           .sort_values('GDP'))

long = factors.melt(id_vars='Region', value_vars=['GDP', 'Freedom'],
                    var_name='Factor', value_name='Value')

richest = factors.iloc[-1]['Region']
freest = factors.sort_values('Freedom').iloc[-1]['Region']
if richest == freest:
    title = f'{richest} leads on both GDP and freedom'
else:
    title = f'{richest} is the richest region, but {freest} feels the freest'

fig = px.bar(
    long,
    x='Value', y='Region',
    orientation='h',
    color='Factor',
    barmode='group',
    color_discrete_map={'GDP': '#2E75B6', 'Freedom': '#FFC000'},
    title=title,
    labels={'Value': 'Average sub-factor score', 'Region': '', 'Factor': ''},
    height=480
)
fig.update_traces(marker_line_width=0)
fig.update_layout(
    xaxis=dict(rangemode='tozero', gridcolor='#EEEEEE'),   # zero baseline
    yaxis=dict(gridcolor='white',
               categoryorder='array',
               categoryarray=factors['Region'].tolist()),
    plot_bgcolor='white',
    paper_bgcolor='white',
    font=dict(family='Arial', size=13),
    legend=dict(orientation='h', y=1.08, x=0, traceorder='reversed'),
    margin=dict(l=10, r=30, t=80, b=40)
)
fig.show()
```
